```python
import requests
from bs4 import BeautifulSoup
```

```python
r= requests.get("https://quotes.toscrape.com")
c= r.content
```

```python
soup=BeautifulSoup(c, "html.parser")
```

```python
print(soup.prettify())
```

```python
all=soup.find_all("div",{"class":"quote"})
```

```python
all[0]
```

```python
all[0].find_all("small")[0].text
```

```python
for item in all:
    print(item.find_all("small")[0].text)
```
